```python
from pydantic import BaseModel,Field,EmailStr #Next Level validation
from typing import Optional, Literal
class employee_schema(BaseModel):
    name:str = Field(..., min_length=1, max_length=50) #... significa required
    id:int = Field(..., gt=0)
    email:EmailStr = Field(..., description="Employee's email address")
    department:Optional[str] = Field(None, max_length=50) #tambien se puede solo none
    gender:Literal["Male", "Female","Other"] = Field(..., description="Employee's gender")
```

```python
obj = employee_schema(name="John", id=123, email="john@example.com", gender="Male")
obj
```

## Serialization 

```python
obj.model_dump() #dict actual como un diccionario
```

```python
obj.model_dump_json() #json
```
