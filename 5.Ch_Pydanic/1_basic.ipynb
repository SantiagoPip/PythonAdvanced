```python
class Employee_Schema(BaseModel):
    name:str
    id:int
    department:str
obj = Employee_Schema(name="John", id=123, department="IT")
print(obj) #pidantyc obj
obj
```

```python
obj2 = Employee_Schema(**{"name":"John", "id":123, "department":"IT"})
obj3 = Employee_Schema(**{"name":"John", "id":"123", "department":"HR"})
print(obj2)
print(obj3)
```

### StrictTypes

```python
from pydantic import StrictInt, StrictStr
```

```python
class Employee_Schema_Strict(BaseModel):
    name:StrictStr
    id:StrictInt
    department:StrictStr
```

```python
obj4 = Employee_Schema_Strict(**{"name":"John", "id":123, "department":"IT"})
print(obj4)
```

```python
obj5 = Employee_Schema_Strict(**{"name":"John", "id":"123", "department":"IT"})
print(obj5) # This will raise a ValidationError because id is not a StrictInt
```
